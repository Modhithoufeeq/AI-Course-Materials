## 3. What is CrewAI?

**CrewAI** is a framework for orchestrating multi-agent systems where specialized AI agents collaborate to solve complex tasks.

### Why CrewAI?
- **Role-based agents**: Each agent has a specific role, goal, and backstory
- **Task delegation**: Agents can delegate tasks to each other
- **Process types**: Sequential and hierarchical workflows
- **Built-in memory**: Agents remember context across tasks
- **Tool integration**: Easy to add custom tools for agents

### CrewAI Architecture
```
CREW
  Agent 1 (Role)    Agent 2 (Role)    Agent 3 (Role)
       |                  |                  |
  Task 1 ---------> Task 2 ---------> Task 3
```

### Core Concepts
| Concept | Description |
|---------|-------------|
| **Agent** | An AI worker with a role, goal, and backstory |
| **Task** | A specific assignment given to an agent |
| **Crew** | A collection of agents and tasks working together |
| **Process** | How tasks are executed (Sequential or Hierarchical) |
| **Tool** | External capabilities an agent can use |

## 5. Create Your First Agent

An agent needs three key attributes:
- **Role**: Who the agent is (e.g., 'Research Analyst')

## 6. Create Your First Task

A task defines what needs to be done and who should do it.
- **Description**: Detailed instructions for the task
- **Expected Output**: What the final deliverable should look like
- **Agent**: Which agent is responsible

````python
# Create a research taskresearch_task = Task(    description=(        'Research the current state of artificial intelligence in healthcare. '        'Focus on: 1) Key applications and use cases, 2) Major companies and technologies, '        '3) Recent breakthroughs (2024-2025), 4) Challenges and limitations. '        'Provide a comprehensive overview with specific examples.'    ),    expected_output=(        'A detailed report (at least 500 words) covering AI applications in healthcare, '        'including specific companies, technologies, breakthroughs, and challenges. '        'The report should be well-structured with clear sections.'    ),    agent=researcher)print('Research Task created:')print(f'   Description length: {len(research_task.description)} chars')print(f'   Assigned to: {research_task.agent.role}')
````

## 7. Create the Crew

A **Crew** brings agents and tasks together and defines how they execute.
- **Agents**: List of all agents in the crew
- **Tasks**: List of all tasks to execute
- **Process**: `sequential` (default) or `hierarchical`
- **Verbose**: Show detailed execution logs

````python
# Create the crewmy_crew = Crew(    agents=[researcher],    tasks=[research_task],    process=Process.sequential,    verbose=True,    memory=False,    max_rpm=10)print('Crew created successfully!')print(f'   Agents: {len(my_crew.agents)}')print(f'   Tasks: {len(my_crew.tasks)}')print(f'   Process: {my_crew.process}')
````

## 8. Run the Crew

Execute the crew by calling `kickoff()`. This starts the workflow and returns the final output.

````python
# Execute the crewprint('=' * 60)print('KICKING OFF THE CREW')print('=' * 60)result = my_crew.kickoff()print('\n' + '=' * 60)print('CREW EXECUTION COMPLETE')print('=' * 60)
````

## 9. Display the Final Output
