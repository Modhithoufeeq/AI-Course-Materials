## 2. Imports & Setup

```python
from crewai import Agent, Task, Crew, Processfrom langchain_openai import ChatOpenAIimport osfrom dotenv import load_dotenvload_dotenv()llm = ChatOpenAI(    model='gpt-4o-mini',    temperature=0.7,    api_key=os.getenv('OPENAI_API_KEY'))print('LLM configured successfully')
```

## 3. Theory: What is an Agent?

A **CrewAI Agent** is an autonomous AI worker with:

### Core Attributes
| Attribute | Purpose | Example |
|-----------|---------|---------|
| **Role** | Defines who the agent is | 'Senior Data Analyst' |
| **Goal** | What the agent aims to achieve | 'Extract actionable insights from data' |
| **Backstory** | Personality and context | 'You have 10 years of experience...' |

### Configuration Options
| Option | Description |
|--------|-------------|
| `allow_delegation` | Can the agent delegate tasks to others? |
| `verbose` | Show detailed thinking process |
| `max_iter` | Maximum iterations before giving up |
| `tools` | List of tools the agent can use |
| `memory` | Enable agent memory |
| `llm` | The language model to use |

### Agent Specialization
Specialized agents are more effective than general ones because:
- They have domain-specific context (backstory)
- Their goals are narrowly defined
- They produce higher quality output in their domain

## 4. Create a Researcher Agent

The researcher specializes in gathering information and finding sources.

```python
# Researcher Agentresearcher = Agent(    role='Senior Research Analyst',    goal='Discover and synthesize comprehensive information on any topic with high accuracy',    backstory=(        'You are a world-class research analyst with 15 years of experience in academic and '        'industry research. You excel at finding reliable sources, cross-referencing information, '        'and presenting findings in a structured manner. You are meticulous, thorough, and always '        'cite your sources. You have worked with top consulting firms and research institutions.'    ),    llm=llm,    verbose=True,    allow_delegation=False,    max_iter=5)print('Researcher Agent Created:')print(f'  Role: {researcher.role}')print(f'  Goal: {researcher.goal}')print(f'  Allow Delegation: {researcher.allow_delegation}')print(f'  Max Iterations: {researcher.max_iter}')
```

## 5. Create an Analyst Agent

The analyst specializes in interpreting data and extracting insights.

```python
# Analyst Agentanalyst = Agent(    role='Data Insights Analyst',    goal='Transform raw research data into actionable insights and strategic recommendations',    backstory=(        'You are a senior data analyst with expertise in pattern recognition, trend analysis, '        'and strategic thinking. You have worked with Fortune 500 companies to turn complex data '        'into clear business recommendations. You are known for your ability to identify '        'non-obvious connections and provide data-driven insights that drive decision-making.'    ),    llm=llm,    verbose=True,    allow_delegation=False,    max_iter=5)print('Analyst Agent Created:')print(f'  Role: {analyst.role}')print(f'  Goal: {analyst.goal}')
```

## 6. Create a Writer Agent

The writer specializes in creating polished, engaging content.

```python
# Writer Agentwriter = Agent(    role='Professional Content Writer',    goal='Create compelling, well-structured, and engaging content for any audience',    backstory=(        'You are an award-winning content writer with 12 years of experience crafting '        'narratives for blogs, reports, and publications. You have a gift for making complex '        'topics accessible and engaging. Your writing has been featured in major publications '        'and you understand how to adapt tone and style for different audiences.'    ),    llm=llm,    verbose=True,    allow_delegation=False,    max_iter=5)print('Writer Agent Created:')print(f'  Role: {writer.role}')print(f'  Goal: {writer.goal}')
```

## 7. Test Individual Agents

Let's test each agent independently by assigning them a simple task.

```python
# Test Researcherprint('=' * 60)print('TESTING RESEARCHER AGENT')print('=' * 60)research_task = Task(    description='Research the top 3 benefits of renewable energy in 2025. Keep it brief (200 words).',    expected_output='A concise summary of the top 3 benefits of renewable energy.',    agent=researcher)test_crew = Crew(agents=[researcher], tasks=[research_task], verbose=True)result1 = test_crew.kickoff()print('\nResearcher Output:')print(result1)
```

## 8. Test Analyst Agent

```python
print('\n' + '=' * 60)print('TESTING ANALYST AGENT')print('=' * 60)analysis_task = Task(    description=(        'Analyze the following statement and provide key insights: '        'Solar energy adoption has grown by 40% annually since 2022, '        'while costs have dropped by 60%. Battery storage technology has improved by 300%.'    ),    expected_output='Key insights and trends derived from the data provided.',    agent=analyst)test_crew2 = Crew(agents=[analyst], tasks=[analysis_task], verbose=True)result2 = test_crew2.kickoff()print('\nAnalyst Output:')print(result2)
```

## 9. Test Writer Agent

```python
print('\n' + '=' * 60)print('TESTING WRITER AGENT')print('=' * 60)writing_task = Task(    description=(        'Write a short, engaging paragraph (100 words) about the future of AI '        'in everyday life. Make it inspiring and accessible to a general audience.'    ),    expected_output='An engaging, well-written paragraph about AI in everyday life.',    agent=writer)test_crew3 = Crew(agents=[writer], tasks=[writing_task], verbose=True)result3 = test_crew3.kickoff()print('\nWriter Output:')print(result3)
```

## 10. Compare Agent Specializations

Notice how each agent approaches tasks differently based on their role, goal, and backstory:

- **Researcher**: Focuses on facts, sources, and comprehensiveness
- **Analyst**: Focuses on patterns, insights, and implications
- **Writer**: Focuses on narrative, engagement, and readability
