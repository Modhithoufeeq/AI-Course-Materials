## 4. What is a CrewAI Agent?

A CrewAI Agent is an autonomous AI worker with a specialized role.

### Core attributes

| Attribute | Purpose | Example |
|---|---|---|
| Role | Defines who the agent is | Senior Research Analyst |
| Goal | Defines what the agent should achieve | Find accurate research |
| Backstory | Provides context and expertise | 15 years of research experience |

### Important configuration

- `allow_delegation`
- `verbose`
- `max_iter`
- `llm`
- `tools`
- `memory`

Specialized agents are useful because each agent has a focused responsibility.

## 5. Create Researcher Agent

````python
researcher = Agent(
    role="Senior Research Analyst",
    goal=(
        "Discover and synthesize comprehensive information "
        "on any topic with high accuracy"
    ),
    backstory=(
        "You are a world-class research analyst with 15 years "
        "of experience in academic and industry research. "
        "You excel at finding reliable information, "
        "cross-referencing facts, and presenting findings "
        "in a structured manner. You are meticulous, "
        "thorough, and evidence-oriented."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False,
    max_iter=5
)

print("Researcher Agent Created")
print("Role:", researcher.role)
print("Goal:", researcher.goal)
print("Delegation:", researcher.allow_delegation)
print("Max Iterations:", researcher.max_iter)
````

## 6. Create Analyst Agent

````python
analyst = Agent(
    role="Data Insights Analyst",
    goal=(
        "Transform research information into actionable insights "
        "and strategic recommendations"
    ),
    backstory=(
        "You are a senior data analyst with expertise in "
        "pattern recognition, trend analysis, and strategic "
        "thinking. You are experienced in turning complex "
        "information into clear business recommendations. "
        "You focus on evidence, patterns, implications, "
        "and practical decision-making."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False,
    max_iter=5
)

print("Analyst Agent Created")
print("Role:", analyst.role)
print("Goal:", analyst.goal)
print("Delegation:", analyst.allow_delegation)
print("Max Iterations:", analyst.max_iter)
````

## 7. Create Writer Agent

````python
writer = Agent(
    role="Professional Content Writer",
    goal=(
        "Create compelling, well-structured, accurate, "
        "and engaging content for the intended audience"
    ),
    backstory=(
        "You are an experienced professional content writer "
        "with expertise in blogs, reports, and publications. "
        "You make complex topics accessible and engaging. "
        "You adapt tone, structure, and detail based on "
        "the audience and purpose of the content."
    ),
    llm=llm,
    verbose=True,
    allow_delegation=False,
    max_iter=5
)

print("Writer Agent Created")
print("Role:", writer.role)
print("Goal:", writer.goal)
print("Delegation:", writer.allow_delegation)
print("Max Iterations:", writer.max_iter)
````

## 8. Compare the Three Agents

### Researcher
Focuses on:
- Facts
- Research

- Comprehensive findings

### Analyst
Focuses on:

### Writer
Focuses on:

## 9. Test Researcher Agent

````python
print("=" * 60)
print("TESTING RESEARCHER AGENT")
print("=" * 60)

research_task = Task(
    description=(
        "Research the top 3 benefits of renewable energy "
        "and explain why they are important. "
        "Keep the response concise."
    ),
    expected_output=(
        "A concise research summary covering the top 3 "
        "benefits of renewable energy."
    ),
    agent=researcher
)

research_test_crew = Crew(
    agents=[researcher],
    tasks=[research_task],
    process=Process.sequential,
    verbose=True,
    memory=False
)

research_result = await research_test_crew.kickoff_async()

print("Researcher Output:")
print(research_result)
````

## 10. Test Analyst Agent

````python
print("=" * 60)
print("TESTING ANALYST AGENT")
print("=" * 60)

analysis_task = Task(
    description=(
        "Analyze the following business information and "
        "identify the most important insights: "
        "Customer adoption increased by 40%, revenue increased "
        "by 25%, while support tickets decreased by 15%. "
        "Explain the likely business implications."
    ),
    expected_output=(
        "A concise analysis containing key patterns, "
        "business insights, and practical implications."
    ),
    agent=analyst
)

analyst_test_crew = Crew(
    agents=[analyst],
    tasks=[analysis_task],
    process=Process.sequential,
    verbose=True,
    memory=False
)

analysis_result = await analyst_test_crew.kickoff_async()

print("Analyst Output:")
print(analysis_result)
````

## 11. Test Writer Agent

````python
print("=" * 60)
print("TESTING WRITER AGENT")
print("=" * 60)

writing_task = Task(
    description=(
        "Write a short, engaging paragraph about the future "
        "of AI in everyday life. Make it inspiring, clear, "
        "and accessible to a general audience."
    ),
    expected_output=(
        "A polished and engaging paragraph about AI "
        "in everyday life."
    ),
    agent=writer
)

writer_test_crew = Crew(
    agents=[writer],
    tasks=[writing_task],
    process=Process.sequential,
    verbose=True,
    memory=False
)

writing_result = await writer_test_crew.kickoff_async()

print("Writer Output:")
print(writing_result)
````

## 12. Inspect Agent Configuration

````python
print("=" * 60)
print("AGENT CONFIGURATION SUMMARY")
print("=" * 60)

agents = [researcher, analyst, writer]

for agent in agents:
    print(f"\nAgent: {agent.role}")
    print(f"Goal: {agent.goal}")
    print(f"Delegation: {agent.allow_delegation}")
    print(f"Max Iterations: {agent.max_iter}")
````
